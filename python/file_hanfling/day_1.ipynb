```python
big_L = ['hello world\n' for i in range(1000)]
with open('files/big.txt','w') as f:
    f.writelines(big_L)
```

```python
with open("files/big.txt", 'r') as f:
    s = f.read()
    print(s)
```

```python
with open("files/big.txt", "r") as f:
    chunk_size = 10

    while True:
        chunk = f.read(chunk_size)
        if not chunk:  # An empty string "" signals End of File (EOF)
            break
        print(chunk, end="")
```
