```python
def odd_even():
    even = []
    odd = []

    for i in range(1,101):
        if i%2 == 0:
            even.append(i)
        else:
            odd.append(i)
    print("Even numbers: ", even)
    print("Odd numbers: ", odd)

odd_even()
```

```python
def power(a=2,b=3):
    return a**b    #a^b

power(b=4,a=3)
```

```python
def multiply(a,b,c,d):
    return a*b*c*d

multiply(1,2,3,4)
```

```python
def multiply(*args):
    product = 1
    print(type(args))
    for i in args:
        product = product * i
    return product
```

```python
multiply(1,2,3,4,5,6,7,8,9)
```

```python
def display(**kwargs):
    print(type(kwargs))
    for(key,value) in kwargs.items():
        print(f"{key} -> {value}") 
```

```python
display(india="Delhi", Bangladesh="Dhaka",srilanka="colombo")
```
