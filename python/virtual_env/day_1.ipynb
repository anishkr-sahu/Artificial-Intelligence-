Virtual Environment:
- A private, separate workspace on your computer for a specific coding project on a specific version of a programming language.

uses:
- avoids version conflict
- keeps computer clean
- Easy sharing

open jupyter notebook
- cmd: jupyter notebook

Virtual Environment create:
- conda create -n project_name python=3.14 -y
- conda activate project_name
- pip list
- pip install pandas==
- conda deactivate
- conda env list
- conda activate base
- conda remove -n project_name --all

requirements.txt
- a file which contain all the necessary libraries which have to install in the project
- pip install -r requirements.txt
