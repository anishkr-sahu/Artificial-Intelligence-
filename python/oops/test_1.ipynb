```python
class AtmMachine:
    tech_name = "ASML"  # class variable
    __customer_id = 0
    def __init__(self):
        self.pin = ""
        self.__balance = 0
        self.customer_id = AtmMachine.__customer_id
        #self.menu()
    @staticmethod
    def get_customer_id():
        return AtmMachine.__customer_id
    
    def get_balance(self):
        return self.__balance

    def set_balance(self,new_balance):
        if type(new_balance) == int:
            self.__balance = new_balance
        else:
            print("error")


    def menu(self):
        user_input = input(
            """
            Hi how can i help you ?

            1. Press 1 to create pin
            2. Press 2 to change pin
            3. press 3 to check balance
            4. press 4 to withdraw
            5. Anything to exit
            """
        )

        if user_input == "1":
            self.create_pin()
        elif user_input == "2":
            self.change_pin()
        elif user_input == "3":
            self.check_balance()
        elif user_input == "4":
            self.withdraw_balance()
        else:
            exit()

    def create_pin(self):
        user_pin = input("Enter your pin: ")
        self.pin = user_pin

        user_balance = int(input("Enter balance: "))
        self.__balance = user_balance

        print("Pin Created Successfully!")
        #self.menu()

    def change_pin(self):
        old_pin = input("Enter your pin: ")

        if old_pin == self.pin:
            new_pin = input("Enter the new pin: ")
            self.pin = new_pin
            print('Pin changed Successfully!')
            self.menu()
        else:
            print('Invalid pin!')
            #self.menu()

    def check_balance(self):
        user_pin = input("Enter your pin: ")
        if user_pin == self.pin:
            print(f"Your balance is {self.__balance}")
        else:
            print("Your pin is incorrect, please try again")
            #self.menu()

    def withdraw_balance(self):
        user_pin = input("Enter the pin: ")
        if user_pin == self.pin:
            amount = int(input("Enter the amount: "))
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"You have withdraw {amount}. your new balance is {self.__balance}")
            else:
                print("Insufficient balance!")
        else:
            print("Your pin is incorrect, please try again")
            #self.menu()
```

```python
class MathUtils:
    
    @staticmethod
    def is_even(number):
        """A utility function that checks if a number is even."""
        return number % 2 == 0

    @staticmethod
    def add(a, b):
        """A utility function to add two numbers."""
        return a + b


# 1. Calling static methods directly on the CLASS (Recommended approach)
print(MathUtils.is_even(4))  # Output: True
print(MathUtils.is_even(7))  # Output: False

print(MathUtils.add(10, 5))   # Output: 15

# 2. You can also call them on an INSTANCE (Though less common)
utils_obj = MathUtils()
print(utils_obj.is_even(8))   # Output: True
```

```python
# polymorphism : many form it means using exact same method name for different types of object 
# and having each object respond in its own way

class Shape:
    def area(self,radius,l,b):
        if b == None:
            return 3.14*radius**2
        else:
            return l*b

s = Shape()
result = s.area(radius =4, l=None, b=None)
print(result) 
```

```python
# abstract : abc
from abc import ABC, abstractmethod

class BankServer(ABC):
    def database(self):
        print("Connected to database")
    def login(self):
        print("Connected to login")

    @abstractmethod   # hr child class ko compulsary hai yae security method create krna
    def security(self):
        print("Security passed")

class MobileApp(BankServer):
    def mobile_login(self):
        print("Mobile app login successfully")

    def security(self):
        print("Mobile app security")
```
